**Data dari Exercise 3**

```python
meals = [
    "Biss Jaggery Modak",
    "Honey Cookies",
    "Gingerbread Waffles",
    "Sauerkraut and Fish Salad",
    "Carrot and Bean Fritters",
    "Apple Pie",
    "Quark and Rhubarb Pie",
    "Papaya salad"
]

ingredients = [
    "Chicken",
    "Salmon",
    "Beef",
    "Pork"
]

print("=== Latest Meals ===")
for i, meal in enumerate(meals, 1):
    print(i, ".", meal)

print("\n=== Popular Ingredients ===")
for i, ingredient in enumerate(ingredients, 1):
    print(i, ".", ingredient)
```

**2.1 Mengubah Teks Menjadi Lowercase**

Tahap pertama adalah mengubah seluruh huruf pada teks menjadi huruf kecil agar penulisan teks menjadi lebih seragam.

```python
meals_lower = [meal.lower() for meal in meals]
ingredients_lower = [ingredient.lower() for ingredient in ingredients]

print("=== Latest Meals Setelah Lowercase ===")
for i, meal in enumerate(meals_lower, 1):
    print(i, ".", meal)

print("\n=== Popular Ingredients Setelah Lowercase ===")
for i, ingredient in enumerate(ingredients_lower, 1):
    print(i, ".", ingredient)
```

**2.2 Menghapus Tanda Baca dan Karakter yang Tidak Diperlukan**

Pada tahap ini, tanda baca dan karakter yang tidak diperlukan dihapus dari teks. Huruf, angka, dan spasi tetap dipertahankan.

```python
import re

meals_clean = [
    re.sub(r'[^a-z0-9\s]', '', meal)
    for meal in meals_lower
]

ingredients_clean = [
    re.sub(r'[^a-z0-9\s]', '', ingredient)
    for ingredient in ingredients_lower
]

print("=== Latest Meals Setelah Pembersihan ===")
for i, meal in enumerate(meals_clean, 1):
    print(i, ".", meal)

print("\n=== Popular Ingredients Setelah Pembersihan ===")
for i, ingredient in enumerate(ingredients_clean, 1):
    print(i, ".", ingredient)
```

**2.3 Menghapus Stopwords**

Stopwords merupakan kata-kata umum yang biasanya memiliki informasi relatif rendah dalam pengolahan teks. Karena data TheMealDB menggunakan bahasa Inggris, digunakan stopwords bahasa Inggris.

```python
import nltk

nltk.download('stopwords')

from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

meals_no_stopwords = []

for meal in meals_clean:
    words = meal.split()
    filtered_words = [word for word in words if word not in stop_words]
    meals_no_stopwords.append(" ".join(filtered_words))

ingredients_no_stopwords = []

for ingredient in ingredients_clean:
    words = ingredient.split()
    filtered_words = [word for word in words if word not in stop_words]
    ingredients_no_stopwords.append(" ".join(filtered_words))

print("=== Latest Meals Setelah Stopwords Dihapus ===")
for i, meal in enumerate(meals_no_stopwords, 1):
    print(i, ".", meal)

print("\n=== Popular Ingredients Setelah Stopwords Dihapus ===")
for i, ingredient in enumerate(ingredients_no_stopwords, 1):
    print(i, ".", ingredient)
```

**2.4 Menghilangkan Spasi Berlebih**

Tahap ini dilakukan untuk menghilangkan spasi yang berlebihan sehingga setiap kata dipisahkan menggunakan satu spasi.

```python
meals_final = [" ".join(meal.split()) for meal in meals_no_stopwords]
ingredients_final = [" ".join(ingredient.split()) for ingredient in ingredients_no_stopwords]

print("=== Latest Meals ===")
for i, meal in enumerate(meals_final, 1):
    print(i, ".", meal)

print("\n=== Popular Ingredients ===")
for i, ingredient in enumerate(ingredients_final, 1):
    print(i, ".", ingredient)
```

**2.5 Hasil Akhir Preprocessing**

Bagian ini menampilkan hasil akhir preprocessing pada data *latest Meals* dan *Popular Ingredients*

```python
print("========== HASIL PREPROCESSING ==========")

print("\n--- Latest Meals ---")

for i, (before, after) in enumerate(zip(meals, meals_final), 1):
    print(f"{i}. Sebelum : {before}")
    print(f"   Sesudah : {after}")

print("\n--- Popular Ingredients ---")

for i, (before, after) in enumerate(zip(ingredients, ingredients_final), 1):
    print(f"{i}. Sebelum : {before}")
    print(f"   Sesudah : {after}")
```
