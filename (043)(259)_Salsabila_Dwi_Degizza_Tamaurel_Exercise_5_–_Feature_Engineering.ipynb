**Exercise 5 — Feature Engineering Part 1**



```
1. One-Hot Encoding
Pada tahap ini, data teks hasil preprocessing diubah menjadi bentuk numerik menggunakan One-Hot Encoding. Metode yang digunakan adalah `CountVectorizer` dengan `binary=True`, sehingga setiap kata direpresentasikan dengan nilai 1 jika muncul pada teks dan 0 jika tidak muncul.
```



````python
import pandas as pd

df = pd.read_csv("hasil_preprocessing_exercise_4.csv")

df
````

````python
# Combine the non-null elements of both 'meals_cleaned' and 'ingredients_cleaned' columns
meals = df["meals_cleaned"].dropna().tolist()
ingredients = df["ingredients_cleaned"].dropna().tolist()

documents = meals + ingredients
documents
````

````python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

# Mengubah teks menjadi numerik dengan One-Hot Encoding
vectorizer = CountVectorizer(binary=True)

X = vectorizer.fit_transform(documents)

df_onehot = pd.DataFrame(
    X.toarray(),
    columns=vectorizer.get_feature_names_out()
)

df_onehot
````



```
2. TF-IDF
Pada tahap ini, data teks diubah menjadi matriks TF-IDF menggunakan `TfidfVectorizer`. TF-IDF memberikan bobot pada kata berdasarkan kemunculannya dalam suatu dokumen dan seberapa umum kata tersebut pada seluruh dokumen.
```



````python
from sklearn.feature_extraction.text import TfidfVectorizer

# Mengubah teks menjadi matriks TF-IDF
vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(documents)

df_tfidf = pd.DataFrame(
    X.toarray(),
    columns=vectorizer.get_feature_names_out()
)

df_tfidf
````



```
3. Applied to the Preprocessed Data
Pada tahap ini, One-Hot Encoding dan TF-IDF diterapkan pada data teks yang telah melalui preprocessing pada Exercise 4. Data yang digunakan berasal dari hasil preprocessing nama makanan pada **Latest Meals** dan bahan pada **Popular Ingredients** dari TheMealDB.
```



````python
df_preprocessed = pd.DataFrame({
    "prepro": documents
})

df_preprocessed
````

````python
vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(df_preprocessed["prepro"])

df_FE = pd.DataFrame(
    X.toarray(),
    columns=vectorizer.get_feature_names_out()
)

df_FE
````



>Analisis Hasil

Berdasarkan hasil One-Hot Encoding, data teks berhasil diubah menjadi bentuk numerik berupa nilai 0 dan 1. Nilai 1 menunjukkan bahwa suatu kata terdapat pada dokumen, sedangkan nilai 0 menunjukkan bahwa kata tersebut tidak terdapat pada dokumen.

Pada hasil TF-IDF, setiap kata memiliki bobot yang berbeda. Bobot tersebut menunjukkan tingkat kepentingan suatu kata dalam dokumen dibandingkan dengan keseluruhan dokumen. Dengan demikian, hasil feature engineering dapat digunakan sebagai representasi numerik dari data teks hasil preprocessing.



**Exercise 5 - Feature Engineering Part 2**

Bagian ini mengerjakan Part 2 menggunakan data hasil preprocessing dari Exercise 4.

````python
import pandas as pd

df = pd.read_csv("hasil_preprocessing_exercise_4.csv")

df.head()
````

````python
meals = df['meals_cleaned'].fillna('').astype(str).tolist()
ingrediens = df['ingredients_cleaned'].fillna('').astype(str).tolist()
documents = [m + '' + ing for m, ing in zip(meals, ingrediens)]
documents = [doc.strip() for doc in documents if doc.strip()]

print('Jumlah dokumen:', len(documents))
print('Contoh dokumen:', documents[0] if documents else 'Data Kosong')
````

Bagian ini menerapkan tokenisasi, Word2Vec, dan FastText pada data hasil preprocessing yang sama.

## Tokenize Sentence

````python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')

from nltk import word_tokenize
from tqdm.auto import tqdm

sentences = [word_tokenize(str(doc).lower()) for doc in tqdm(documents)]
sentences[:5]
````

# Build Word2Vec Model

````python
!pip -q install gensim
````

````python
from gensim.models import Word2Vec
import numpy as np
import pandas as pd

model_w2v = Word2Vec(sentences, min_count=1, vector_size=100, window=5, sg=0)

w2v = model_w2v.wv

print("10 kata pertama:", w2v.index_to_key[:10])
print("Ukuran matriks vektor kata:", w2v.vectors.shape)
print("Dimensi setiap vektor:", w2v.vector_size)
````

````python
if w2v.index_to_key:
  kata_contoh = w2v.index_to_key[0]
  print("Kata contoh:", kata_contoh)
  print("Kata yang mirip:", w2v.similar_by_key(kata_contoh, topn=5))
````

````python
def get_documents_vector(tokens, keyed_vectors):
  valid_vectors = [keyed_vectors[word] for word in tokens if word in keyed_vectors]
  if not valid_vectors:
    return np.zeros(keyed_vectors.vector_size)
  return np.mean(valid_vectors, axis=0)

  X_word2vec = np.array([get_documents_vector(tokens, w2v) for tokens in tqdm(sentences)])
  df_word2vec = pd.DataFrame(X_word2vec)

  display(df_word2vec.head())
  print("Ukuran matriks vektor dokumen Word2Vec:", X_word2vec.shape)
````

# Visualisasi Word2Vec

````python
!pip -q install umap-learn plotly
````

````python
from umap import UMAP
import plotly.express as px

if len(w2v.index_to_key) >= 3:
  n_neighbors = min(15, len(w2v.index_to_key) - 1)
  coords = UMAP(n_components=2, random_state=42, n_neighbors=n_neighbors).fit_transform(w2v.vectors)

  df_umap_w2v = pd.DataFrame(coords, columns=["umap1", "umap2"])
  df_umap_w2v["word"] = w2v.index_to_key

  fig = px.scatter(df_umap_w2v, x="umap1", y="umap2", text="word", title="Word2Vec Visualization")
  fig.update_traces(textposition = "top center")
  fig.update_layout(height=700)
  fig.show()
else:
  print("Jumlah kata terlalu sedikit untuk visualisasi UMAP.")
````

# Build FastText Model

````python
from gensim.models import FastText

model_fasttext = FastText(sentences, min_count=1, vector_size=100, window=5, min_n=3, max_n=6, sg=1, epochs=10)

fasttext = model_fasttext.wv

print("10 kata pertama:", fasttext.index_to_key[:10])
print("Ukuran matriks vektor kata:", fasttext.vectors.shape)
print("Dimensi setiap vektor:", fasttext.vector_size)
````

````python
if fasttext.index_to_key:
  kata_contoh_fasttext = fasttext.index_to_key[0]
  print("Kata contoh:", kata_contoh_fasttext)
  print("Kata yang mirip:", fasttext.similar_by_key(kata_contoh_fasttext, topn=5))
````

````python
X_fasttext = np.array([get_documents_vector(doc, fasttext) for doc in sentences])
df_fasttext = pd.DataFrame(X_fasttext)

display(df_fasttext.head())
print("Ukuran matriks vektor dokumen FastText:", X_fasttext.shape)
````

# Visualisasi FastText

````python
if len(fasttext.index_to_key) >= 3:
  n_neighbors = min(15, len(fasttext.index_to_key) - 1)
  coords_fasttext = UMAP(n_components=2, random_state=42, n_neighbors=n_neighbors).fit_transform(fasttext.vectors)

  df_umap_fasttext = pd.DataFrame(coords_fasttext, columns=["umap1", "umap2"])
  df_umap_fasttext["word"] = fasttext.index_to_key

  fig = px.scatter(df_umap_fasttext, x="umap1", y="umap2", text="word", title="FastText Visualization")
  fig.update_traces(textposition = "top center")
  fig.update_layout(height=700)
  fig.show()
else:
  print("Jumlah kata terlalu sedikit untuk visualisasi UMAP.")
````

# Analisis Hasil Part 2

Word2Vec dan FastText menghasilkan representasi teks dalam bentuk vektor berdimensi 100. Pada Word2Vec, vektor kata dipelajari berdasarkan konteks kata disekitarnya. FastText juga memanfaatkan potongan katakter(character n-grams) sehingga mempertimbangkan struktur subkata. Vektor dokumen diperoleh dengan menghitung rata-rata vektor kata yang terdapat di dalam setiap dokumen.

Pada visualisasi UMAP, kata-kata yang letaknya berdekatan memiliki representasi vektor yang relatif berdekatan menurut model. Namun, kedekatan tersebut tidak selalu berarti kata-kata itu benar-benar sinonim. Karena data yang digunakan berupa nama makanan dan bahan, hasil kemiripan kata perlu ditafsirkan berdasarkan konteks data tersebut. Kualitas hasil juga dipengaruhi oleh jumlah dan variasi teks yang tersedia
